```python
import os
import re
from collections import defaultdict

folder_path = "documents"

inverted_index = defaultdict(set)
vocabulary = set()

for i in range(1, 11):
    filename = f"doc{i}.txt"
    filepath = os.path.join(folder_path, filename)

    with open(filepath, "r", encoding="utf-8") as file:
        text = file.read().lower()

        words = re.findall(r'\b[a-z]+\b', text)

        vocabulary.update(words)

        for word in set(words):
            inverted_index[word].add(filename)

print("Total Documents:", 10)
print("Vocabulary Size:", len(vocabulary))

print("\nSample Inverted Index:\n")

count = 0
for word in sorted(inverted_index):
    print(f"{word} -> {sorted(inverted_index[word])}")
    count += 1
    if count == 30:
        break
```

```python
import re
from collections import defaultdict

# 10 Sample Documents
documents = {
    "doc1": "Artificial intelligence is transforming modern technology. Machine learning helps computers learn from data.",
    "doc2": "Python is a popular programming language. It is widely used in data science and web development.",
    "doc3": "Information retrieval helps users search documents. Search engines use inverted indexes.",
    "doc4": "Cloud computing provides scalable computing resources. Many companies use cloud platforms.",
    "doc5": "Data structures improve the efficiency of algorithms. Trees graphs and hash tables are useful.",
    "doc6": "Natural language processing enables computers to understand text. Chatbots use NLP techniques.",
    "doc7": "Cyber security protects systems from digital attacks. Encryption improves security.",
    "doc8": "Database management systems store and organize information. SQL retrieves records.",
    "doc9": "Deep learning uses neural networks. It is useful for image recognition.",
    "doc10": "Computer networks connect devices. The internet is the largest network."
}

# Create Inverted Index
inverted_index = defaultdict(list)

for doc_name, text in documents.items():

    # Convert to lowercase and extract words
    words = re.findall(r'\b[a-z]+\b', text.lower())

    # Add unique words to index
    for word in set(words):
        inverted_index[word].append(doc_name)

# Print Inverted Index
print("INVERTED INDEX\n")

for word in sorted(inverted_index):
    print(f"{word} -> {inverted_index[word]}")
```
